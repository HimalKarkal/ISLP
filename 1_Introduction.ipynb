# 1. Introduction

Chapter 1 of the book has no problem set. It simply introduces the subject of Statistical Learning using several real world examples like estimating wages, predicting stock prices, and cancer detection. These examples do a great job of presenting the major classes of statistical learning models: supervised/unsupervised and regression/classification. It then provides a brief but comprehensive history of the subject describing early challenges such as limited computational power.

The book's goal, as described by the authors, is to make statistical learning concepts more widely accessible to non-academic circles. They emphasise that it does not replace their earlier, more technical book on the subject called Elements of Statistical Learning. They state four premises on which ISL is based:
1. Statistical learning methods are widely relevant in both academic and non-academic applications,
2. Statistical learning should not be viewed as a series of black boxes,
3. A very deep understanding and ability to build models from the ground up is not necessary for most practitioners,
4. An assumption that the reader is interested in applying statistical learning to solve real world problems

Some basic mathematical notation is presented along with a discussion of the book's organisation. Finally, the datasets used in the book are introduced. These can be downloaded from the [book's website](https://www.statlearning.com).

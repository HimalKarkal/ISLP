# 1. Introduction

Chapter 1 of the book has no problem set. It introduces the subject of Statistical Learning using several real world examples like estimating wages, predicting stock prices, and cancer detection. These examples do a great job of presenting the major classes of statistical learning models: supervised/unsupervised and regression/classification. It then provides a brief but comprehensive history of the subject describing early challenges such as limited computational power.

The book's goal, as described by the authors, is to make statistical learning concepts more widely accessible to non-academic circles. They emphasise that it does not replace their earlier, more technical book on the subject called Elements of Statistical Learning. They state four premises on which ISL is based:
1. Statistical learning methods are widely relevant in both academic and non-academic applications,
2. Statistical learning should not be viewed as a series of black boxes,
3. A very deep understanding and ability to build models from the ground up is not necessary for most practitioners,
4. An assumption that the reader is interested in applying statistical learning to solve real world problems

Some basic mathematical notation is presented along with a discussion of the book's organisation. Finally, the datasets used in the book are introduced. These can be downloaded from the [book's website](https://www.statlearning.com).

## Notes
<ul>
    <li> Statistical Learning refers to a vast set of tools for understanding data. 
    <li>These tools can be classified as supervised or unsupervised. <li> Supervised statistical learning involves building a statistical model for prediction or estimation of an output variable of interest using one or more input variables. The model is created by allowing the output variable to supervise the model's prediction performance.
    <li> Unsupervised statistical learning has input variables but no output variables. It is used to uncover relationships and structure from data.
    <li> Regression involves predicting a continuous or quantitative output value (wage; grade; VO2max; energy expenditure).
    <li> Classification involves predicting a categorical or qualitative output (win/loss; pass/fail; high_risk/low_risk)
    <li> Clustering is a form of unsupervised learning whose goal it is to group data points based on their observed characteristics. There are input variables but no output variable of interest. 
    <li> The least squares method was developed at the beginning of the nineteenth century and first applied to problems in astronomy. 
    <li> In the 1930s and 1940s classification methods like linear discriminant analysis and logistic regression were developed. 
    <li> The Generalised Linear Model (GLM) was developed in the early 1970s. This method described an entire class of statistical learning methods. Linear and logistic regression were both special cases of the GLM.
    <li> Due to advances in computing science in the 1980s, fitting non-linear methods were no longer computationally prohibitive. As a result, classification and regression trees were developed, followed shortly by Generalised Additive Models.
    <li> Neural Networks gained popularity at around the same time.
    <li> In the 1990s, Support Vector Machines (SVMs) emerged.
    <li> Statistical learning has evolved into a specialised subfield of statistics and computer science.

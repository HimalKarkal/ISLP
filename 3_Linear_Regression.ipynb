## Notes

Questions that Linear Regression can provide answers for:
<ul>
    <li> Is there a relationship between the outcome variable and explanatory variables?
    <li> What is the strength of the relationship if one exists?
    <li> Are all explanatory variables related to the outcome variable? Are some related and others unrelated?
    <li> What is the individual strength of relationship between each explanatory variable and the oucome variable?
    <li> How accurately can the outcome variable be predicted in the future?
    <li> Is the relationship between explanatory and outcome variables truly linear? Or can it be transformed to make it linear?
    <li> Is there an interaction or synergy effect among explanatory variables?

### Question 1

**Intercept**: The null hypothesis is that the intercept for the linear model equals 0. Since the p value is extremely small, we can safely reject the null hypothesis and accept that the intercept is not 0.

**TV**: The null hypothesis is that, with newspaper and radio budgets held constant, a change in TV budget does not affect sales. The small p value gives us enough confidence to reject the null hypothesis.

**radio**: The null hypothesis is that, with TV and newspaper budgets held constant, a change in radio budget does not affect sales. Once again, the small p value gives us enough confidence to reject the null hypothesis.

**newspaper**: The null hypothesis is that, with TV and radio budgets held constant, a change in newspaper budget does not affect sales. In this case, the p value is extremely large. Hence, we fail to reject the null hypothesis.

### Question 2

**KNN Classifier**: Its goal is to classify an outcome. Given a value for the parameter k, the algorithm identifies the k data points in the training dataset that are closest to the point in question. The point is then classified to the majority classification among the k nearest neighbours. A smaller k results in a more flexible model (low bias, high variance).

**KNN Regressor**: Its goal is to predict a quantitative value. Given a value for the parameter k, the algorithm identifies the k data points in the training dataset that are closest to the point in question. Then the average output value of the k nearest neighbours is used as the prediction for the point. A smaller k results in a more flexible model (low bias, high variance)

### Question 4

(a) We would expect the RSS to be lower for the cubic regression. This is because the cubic model has more degrees of freedom and overfits the data. It starts modelling the random error in each data point.

(b) We would expect RSS to be lower for the linear regression. This is because, unlike the cubic case, the linear model does not overfit the data. It is likely going to be a be a good estimate for future data. The cubic model overfits on the training data and hence performs extremely well there. However, when presented with new data, it likely will not perform as well.

(c) The training RSS for the cubic regression will be lower. Since the underlying relationship is not linear, the linear models lacks the required degrees of freedom for a good fit. Since the cubic model has more degrees of freedom than the linear model, it will perform better.

(c) The test RSS will likely be better for the cubic model. If the true relationship is cubic or further, then the cubic model will do better. However, if the true relationship is quadratic, then either model could be the better performer. It comes down to whether the bias of the linear model is more impactful than the variance of the cubic model.

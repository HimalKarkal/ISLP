Questions that Linear Regression can provide answers for:
* Is there a relationship between the outcome variable and explanatory variables?
* What is the strength of the relationship if one exists?
* Are all explanatory variables related to the outcome variable? Are some related and others unrelated?
* What is the individual strength of relationship between each explanatory variable and the oucome variable?
* How accurately can the outcome variable be predicted in the future?
* Is the relationship between explanatory and outcome variables truly linear? Or can it be transformed to make it linear?
* Is there an interaction or synergy effect among explanatory variables?

### Simple Linear Regression

* A quantitative response variable Y is predicted using a single explanatory variable X. The true relationship between X and Y is assumed to be approximately linear. Mathematically, this assumed relationship is represented as, $$ Y \approx \beta_0 + \beta_1 X $$

* $\beta_0$ and $\beta_1$ are model coefficients or parameters. The represent the intercept and slope terms of the linear relationship between X and Y. 

* The process of training a simple linear regression model involves generating estimates for these coefficients. Since they are estimates and not the true coeffificients, they are denoted by $\hat{\beta_0}$ and $\hat{\beta_1}$. These can be used to predict Y using future measurements of the predictor variable X. Predictions made in such a manner, on the basis of a particular value for X, are denoted as $\hat{y}$. The model used for prediction is, $$\hat{y} = \hat{\beta_0} + \hat{\beta_1}x$$

* *Least squares* is the most common approach for fitting linear regression models. It provides a measure of closeness between the line generated by the coefficients and the response variables in the training data. 

* Gilbert Strang's MIT-OCW 18.06 lectures on YouTube and his book *Linear Algebra & Its Applications* are excellent resources to understand the linear algebra perspective of least squares. It boils down to projecting points onto a subspace. It is a lot easier to visualise compared to the calculus point of view. 

* In the calculus perspective, our objective is the minimisation of the residual sum of squares (RSS). This is done by selecting optimal values for $\hat{\beta_0}$ and $\hat{\beta_1}$ For a training data set with n explanatory-outcome variable pairs, RSS is mathematically represented as, $$ RSS = \sum_{i = 1}^{n} [y_i - (\hat{\beta_0} + \hat{\beta_1} x_i)]^2$$

* The coefficients obtained through RSS minimisation are called the least squares coefficient estimates. This is analogous to using a sample mean to estimate the population mean.

### Assessing the Accuracy of the Coefficient Estimates

* An unbiased estimator does not systematically over or underestimate the true parameter. If we estimate coefficients on the basis of one dataset, they will not be exactly right. However, if we average the coefficients obtained over a huge number of datasets then the average will be spot on!

* The standard error (SE) is a measure of how far away from the true coefficient a single estimate might be.

* Standard errors can be used to compute confidence intervals. A 95% confidence interval is defined as a range of values such that with 95% probability, the range will contain the true unknown value of the parameter.

* For linear regression, the 95% confidence interval takes the form: $$\hat{\beta} \pm 2 \cdot SE(\hat{\beta}) $$

* Standard Errors can also be used to perform hypothesis tests on the coefficients. The most common hypothesis test involves testing the null hypothesis of: $$ H_0: \text{There is no relationship between X and Y} $$ versus the alternative hypothesis $$ H_a: \text{There is some relationship between X and Y} $$ 

* Mathematically, this corresponds to testing $ H_0: \beta_1 = 0 $ versus $ H_1: \beta_1 \ne 0$

* We need to determine whether our estimate for $\beta_1$ is sufficiently far from zero that we can be confident that $\beta_1$ is non-zero. If the standard error of our estimate $\hat{\beta_1}$ is small, even relatively small values may give us enough confidence to reject the null hypothesis. On the other hand, if the standard error is large, we need $\hat{\beta_1}$ to be large in absolute value to reject the null hypothesis. 

* In practice, we compute the t-statistic, given by $$ t = \frac{\hat{\beta_1} - 0}{SE(\hat{\beta_1})} $$

* The t-statistic measures the number of standard deviations that $\hat{\beta_1}$ is away from 0. Using the t-distribution, one can easily calculate the probability of observing an absolute t-statistic equal to or larger than the one we currently observe. This probability is called the **p-value**. 

* A small p-value indicates that we are unlikely to observe such a substantial association between the predictor and the response due to chance, in the absence of any real association between the predictor and the response. 

* Typical p-value cutoffs are 5% or 1%. If we observe a p-value less than the cutoff, we have enough confidence to reject the null hypothesis and operate as though a relationship exists between X and Y. While we are not entirely sure, we can be quite confident that we are right.

### Assessing the Accuracy of the Model


```python
# EXPERIMENT 1
import numpy as np
rng = np.random.default_rng(seed = 42)

# True function: f = 2 + 3X + epsilon

# Generating epsilon vector
epsilon_vector = rng.normal(0,5,100)

# Generating 100 random inputs between 0 and 10
x = rng.integers(0,11,100)

# Getting true output values
y = np.full(100, 2) + 3 * x + epsilon_vector

# Trying to estimate true function f using linear regression (linear algebra approach)
aug_matrix = np.column_stack((np.ones(100), x))
estimated_coefficients = np.linalg.inv(aug_matrix.T @ aug_matrix) @ aug_matrix.T @ y

# Printing estimated function f'(x)
print(f"f'(X) = {round(estimated_coefficients[0], 4)} + {round(estimated_coefficients[1], 4)} X")
```

Given a value for the parameter k, the algorithm identifies the k data points in the training dataset that are closest to the point in question.
